```python
import tp
import numpy as np
import matplotlib.pyplot as plt
```

```python
with open("CarsonellaRuddii.txt") as genome:
    genome = "".join(genome.read().splitlines())

    # just_k_lengths_attempted=range(140,160)
    
    # success=[]
    # label=[]
    # for x in just_k_lengths_attempted:
    #     lable.append(i)
    #     success.append(run_test_on_genome(genome,x))
    
    # if all(success):
    #     print("HAY QUE BAJAR EL MINIMO DE just_k_lengths_attempted")
        
k_start_tries = 5  # todo setear en 0
k_max_tries = 80  # todo setear en 80
d_start_tries = 0 # TODO setear en 0
d_max_tries = 100 # TODO cambiar a por lo menos 1000

res = np.full((d_max_tries-d_start_tries,k_max_tries-k_start_tries), False)

for k in range(k_start_tries,k_max_tries):
    for d in range(d_start_tries,d_max_tries):
        print(d-d_start_tries+(k-k_start_tries)*(d_max_tries - d_start_tries), "/",(k_max_tries-k_start_tries)*(d_max_tries-d_start_tries))
        try:
            res[d-d_start_tries][k-k_start_tries] = tp.run_test_on_genome(genome,k,d)
        except IndexError:
            pass


```

```python
# Lo unico que genere con IA pq no me importa aprender a usar plt
from matplotlib.colors import ListedColormap
from matplotlib.patches import Patch
colors = ["pink", "#CCFFCC"]
cmap = ListedColormap(colors)

fig, ax = plt.subplots()

ax.imshow(
    res,
    origin="lower",
    aspect="auto",
    interpolation="nearest",
    cmap=cmap
)

plt.title("Intentos de reconstrucción")
# plt.ylabel("longitud de fragmentos de kd_mer")
# plt.xlabel("espacio entre fragmentos de kd_mer")

ax.set_xlabel("longitud de fragmentos de kd_mer")
ax.set_ylabel("espacio entre fragmentos de kd_mer")
ax.set_xticks(
    range(k_max_tries - k_start_tries),
    range(k_start_tries, k_max_tries)
)
ax.set_yticks(
    range(d_max_tries - d_start_tries),
    range(d_start_tries, d_max_tries)
)

# Cell grid
ax.set_xticks(np.arange(-0.5, res.shape[1], 1), minor=True)
ax.set_yticks(np.arange(-0.5, res.shape[0], 1), minor=True)
ax.grid(which="minor", color="gray", linewidth=0.5)
ax.tick_params(which="minor", bottom=False, left=False)

# Legend
ax.legend(handles=[
    Patch(facecolor=colors[1], label="True"),
    Patch(facecolor=colors[0], label="False"),
])

plt.show()
```
